```python
import pickle

# Load the graph coloring data
with open('../temp/graph_coloring_data_29617ex.pkl', 'rb') as f:
    data = pickle.load(f)
len(data[0]['labels'])
```

```python
import random
random.shuffle(data)
```

```python
import numpy as np
import os
import gzip
import torch
root = '/home/jan/projects/CIIRC/column generation/IPM_MPNN/fac6/select_columns'
subfolder = 'sub'
batch_size = 1000
os.makedirs(f'{root}/{subfolder}/raw', exist_ok=True)
ips = []
pkg_idx = 0
for ex in data[:15000]:
    A = np.array(ex['colorings']).transpose()
    labels = np.array(ex['labels'])
    b = np.ones(A.shape[0])
    c = np.zeros(A.shape[1])

    ips.append((torch.from_numpy(A).to(torch.float), torch.from_numpy(b).to(torch.float), torch.from_numpy(c).to(torch.float),labels)) 
    if len(ips) >= batch_size:
            with gzip.open(f'{root}/{subfolder}/raw/instance_{pkg_idx}.pkl.gz', "wb") as file:
                pickle.dump(ips, file)
                pkg_idx += 1
            ips = []

```
